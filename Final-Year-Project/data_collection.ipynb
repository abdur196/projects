```python
import yfinance as yf
import pandas as pd
import ta

# Download some sample data (example: Apple)
data = yf.download("AAPL", start="2020-01-01", end="2023-01-01")

# Ensure 'Close' is always a Series (not a (n,1) DataFrame)
close_prices = data["Close"].squeeze()

# Add RSI (Relative Strength Index)
data["RSI"] = ta.momentum.RSIIndicator(close=close_prices, window=14).rsi()

# Add MACD (Moving Average Convergence Divergence)
macd = ta.trend.MACD(close=close_prices)
data["MACD"] = macd.macd()
data["MACD_Signal"] = macd.macd_signal()
data["MACD_Diff"] = macd.macd_diff()

# Add Bollinger Bands
boll = ta.volatility.BollingerBands(close=close_prices, window=20, window_dev=2)
data["Bollinger_High"] = boll.bollinger_hband()
data["Bollinger_Low"] = boll.bollinger_lband()

# Add 50-day and 200-day Moving Averages
data["MA50"] = close_prices.rolling(window=50).mean()
data["MA200"] = close_prices.rolling(window=200).mean()

print(data.tail())
```
